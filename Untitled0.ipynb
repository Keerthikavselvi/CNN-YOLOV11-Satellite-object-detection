```python
import random
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Define locations
locations = [
    "Sulur", "Karumathampatti", "Madukkarai", "Kuniyamuthur", "Coimbatore North",
    "Karamadai", "Kovaipudur", "Coimbatore South", "Irugur", "Samathur",
    "Saravanampatti", "Valparai", "Mettupalayam", "Saibaba Colony", "Singanallur",
    "Vellalore", "Periyanaickenpalayam", "Pollachi", "Anaimalai", "Gandhipuram",
    "Hopes", "Town hall", "Neelambur"
]

# Generate random IoT sensor data for each location
data = []
for loc in locations:
    temperature = round(random.uniform(20, 40), 2)  # Temperature in Celsius
    air_quality = round(random.uniform(10, 300), 2)  # AQI (Air Quality Index)
    noise_level = round(random.uniform(30, 90), 2)  # dB (Decibels)
    parking_availability = random.randint(0, 100)  # Percentage of available parking spots
    traffic_congestion = round(random.uniform(0, 100), 2)  # Congestion level in %

    data.append([loc, temperature, air_quality, noise_level, parking_availability, traffic_congestion])

# Convert to DataFrame for analysis
df = pd.DataFrame(data, columns=["Location", "Temperature (°C)", "AQI", "Noise Level (dB)", "Parking Availability (%)", "Traffic Congestion (%)"])

# Display dataset
print(df)

# Basic Analysis
print("\nAverage Temperature:", df["Temperature (°C)"].mean())
print("Average AQI:", df["AQI"].mean())
print("Average Noise Level:", df["Noise Level (dB)"].mean())
print("Average Parking Availability:", df["Parking Availability (%)"].mean())
print("Average Traffic Congestion:", df["Traffic Congestion (%)"].mean())

# Identify high-pollution areas
high_pollution = df[df["AQI"] > 100]
print("\nLocations with high air pollution:")
print(high_pollution[["Location", "AQI"]])

# ================= Visualization =================

# Set Seaborn style
sns.set_style("whitegrid")

# Bar Chart: Air Quality Index (AQI) per Location
plt.figure(figsize=(12,6))
sns.barplot(x="Location", y="AQI", data=df, palette="coolwarm")
plt.xticks(rotation=90)
plt.title("Air Quality Index (AQI) per Location")
plt.xlabel("Location")
plt.ylabel("AQI")
plt.show()

# Bar Chart: Noise Levels per Location
plt.figure(figsize=(12,6))
sns.barplot(x="Location", y="Noise Level (dB)", data=df, palette="viridis")
plt.xticks(rotation=90)
plt.title("Noise Pollution Levels per Location")
plt.xlabel("Location")
plt.ylabel("Noise Level (dB)")
plt.show()

# Pie Chart: Parking Availability
plt.figure(figsize=(7,7))
plt.pie(df["Parking Availability (%)"], labels=df["Location"], autopct='%1.1f%%', startangle=140)
plt.title("Parking Availability Across Locations")
plt.show()

# Pie Chart: Traffic Congestion
plt.figure(figsize=(7,7))
plt.pie(df["Traffic Congestion (%)"], labels=df["Location"], autopct='%1.1f%%', startangle=140, colors=sns.color_palette("Set2"))
plt.title("Traffic Congestion Distribution")
plt.show()
```
